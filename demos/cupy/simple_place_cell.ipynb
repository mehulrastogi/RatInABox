```python
import ratinabox
from ratinabox.Environment import Environment
from ratinabox.Agent import Agent
from ratinabox.Neurons import PlaceCells

import gc
```

```python
# THIS IS FOR PROFILING THE CODE : CPU
# CUPY = False
import timeit

def run_simulation():
    Env = Environment()
    Ag = Agent(Env)
    PCs = PlaceCells(Ag, params={"n": 1000000})
    for i in range(int(60 / Ag.dt)):
        Ag.update()
        PCs.update()
    del Env
    del Ag
    del PCs
    gc.collect()

%timeit run_simulation()
```
